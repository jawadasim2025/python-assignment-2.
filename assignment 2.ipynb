```python
number = int(input("Enter a number: "))

total = 0
print("Numbers from 1 to", number, ":")

for i in range(1, number + 1):
    print(i)
    total = total + i

print("The total sum is:", total)

if number % 2 == 0:
    print(number, "is an even number")
else:
    print(number, "is an odd number")

if number > 0:
    print(number, "is a positive number")
elif number < 0:
    print(number, "is a negative number")
else:
    print(number, "is zero")

if number > 0 and number % 2 == 0:
    print("The number is positive and even")
elif number > 0 and number % 2 != 0:
    print("The number is positive and odd")
elif number < 0 and number % 2 == 0:
    print("The number is negative and even")
elif number < 0 and number % 2 != 0:
    print("The number is negative and odd")
```

```python
correct_username = "jawad"
correct_password = "2009"


def login():
    attempts = 3

    while attempts > 0:
        username = input("Enter username: ")
        password = input("Enter password: ")

        if username == correct_username and password == correct_password:
            print("Login successful! Welcome,", username)
            return True
        else:
            attempts = attempts - 1
            print("Wrong username or password.")
            print("Attempts left:", attempts)

    print("Too many wrong attempts. Access denied.")
    return False


print("=== Login System ===")
login()
```

```python
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return "Error: cannot divide by zero"
    else:
        return a / b


print("=== Simple Calculator ===")

choice = "yes"

while choice == "yes":
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    operator = input("Enter operator (+, -, *, /): ")

    if operator == "+":
        result = add(num1, num2)
    elif operator == "-":
        result = subtract(num1, num2)
    elif operator == "*":
        result = multiply(num1, num2)
    elif operator == "/":
        result = divide(num1, num2)
    else:
        result = "Invalid operator"

    print("Result:", result)

    choice = input("Do you want to calculate again? (yes/no): ").lower()

print("Thank you for using the calculator. Goodbye!")
```

```python
balance = 50,000
correct_pin = 2009


def check_balance():
    print("Your current balance is:", balance)


def deposit():
    global balance
    amount = float(input("Enter amount to deposit: "))

    if amount > 0:
        balance = balance + amount
        print("Deposit successful. New balance:", balance)
    else:
        print("Invalid amount. Please enter a positive number.")


def withdraw():
    global balance
    amount = float(input("Enter amount to withdraw: "))

    if amount <= 0:
        print("Invalid amount. Please enter a positive number.")
    elif amount > balance:
        print("Not enough money in your account.")
    else:
        balance = balance - amount
        print("Withdrawal successful. New balance:", balance)


print("=== Welcome to the ATM ===")

attempts = 3
pin_correct = False

while attempts > 0 and not pin_correct:
    pin = int(input("Enter your PIN: "))

    if pin == correct_pin:
        pin_correct = True
    else:
        attempts = attempts - 1
        print("Wrong PIN. Attempts left:", attempts)

if pin_correct:
    running = True

    while running:
        print("\n1. Check balance")
        print("2. Deposit money")
        print("3. Withdraw money")
        print("4. Exit")

        option = input("Choose an option (1-4): ")

        if option == "1":
            check_balance()
        elif option == "2":
            deposit()
        elif option == "3":
            withdraw()
        elif option == "4":
            print("Thank you for using the ATM. Goodbye!")
            running = False
        else:
            print("Invalid option. Please try again.")
else:
    print("Too many wrong attempts. Your card is blocked.")
```
